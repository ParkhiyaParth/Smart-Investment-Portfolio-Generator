```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the data
df1 = pd.read_csv('sample.csv')
df1.drop('Unnamed: 0', axis=1, inplace=True)
df1
```

```python
# Sample extracted data
data = {
    "Close": [180.3086243, 181.771698, 180.5673828, 179.9005432, 178.8165586, 
              174.2770996, 169.3204803, 168.3251953, 168.2057495, 169.9276123]
}
```

```python
df=pd.DataFrame(data)
```

```python
# Number of bins
bins = 3

# Equal Width Binning
min_val, max_val = df["Close"].min(), df["Close"].max()
bin_width = (max_val - min_val) / bins
equal_width_bins = [min_val + i * bin_width for i in range(bins + 1)]
df["Equal Width Bin"] = pd.cut(df["Close"], bins=equal_width_bins, include_lowest=True)

# Equal Frequency Binning
df_sorted = df.sort_values("Close")
df_sorted["Equal Frequency Bin"] = pd.qcut(df_sorted["Close"], bins, labels=False)

# Function to compute smoothed values
def smooth_mean(values):
    return [np.mean(values)] * len(values)

def smooth_median(values):
    return [np.median(values)] * len(values)

def smooth_bound(values):
    return [values[0]] * len(values[:len(values)//2]) + [values[-1]] * len(values[len(values)//2:])

# Compute smoothed values for Equal Frequency Bins
equal_freq_groups = df_sorted.groupby("Equal Frequency Bin")["Close"].apply(list)
smooth_mean_eq_freq = equal_freq_groups.apply(smooth_mean).explode().values
smooth_median_eq_freq = equal_freq_groups.apply(smooth_median).explode().values
smooth_bound_eq_freq = equal_freq_groups.apply(smooth_bound).explode().values

# Compute smoothed values for Equal Width Bins
equal_width_groups = df.groupby("Equal Width Bin")["Close"].apply(list)
smooth_mean_eq_width = equal_width_groups.apply(smooth_mean).explode().values
smooth_median_eq_width = equal_width_groups.apply(smooth_median).explode().values
smooth_bound_eq_width = equal_width_groups.apply(smooth_bound).explode().values

# Prepare output format
output = {
    "Equal Width Binning": equal_width_groups.tolist(),
    "Equal Frequency Binning": equal_freq_groups.tolist(),
    "Smooth Mean (Equal Frequency)": smooth_mean_eq_freq.tolist(),
    "Smooth Mean (Equal Width)": smooth_mean_eq_width.tolist(),
    "Smooth Median (Equal Frequency)": smooth_median_eq_freq.tolist(),
    "Smooth Median (Equal Width)": smooth_median_eq_width.tolist(),
    "Smooth Bound (Equal Frequency)": smooth_bound_eq_freq.tolist(),
    "Smooth Bound (Equal Width)": smooth_bound_eq_width.tolist(),
}

output
```
