```python
import pandas as pd
import numpy as np

# Load the dataset (assuming it's a CSV file, update the filename accordingly)

df = pd.read_csv('sample.csv')
df.drop(['Day','Month','Year'], axis=1, inplace=True)
df.drop(['Unnamed: 0'], axis=1, inplace=True)


# Selecting numerical columns
num_cols = ['Close', 'High', 'Low', 'Open', 'Volume']

# 1. Min-Max Normalization
def min_max_normalization(df, cols):
    min_max_norm = df.copy()
    for col in cols:
        min_val = df[col].min()
        max_val = df[col].max()
        min_max_norm[col] = (df[col] - min_val) / (max_val - min_val)
    return min_max_norm

min_max_df = min_max_normalization(df, num_cols)

# 2. Z-score Normalization
def z_score_normalization(df, cols):
    z_score_norm = df.copy()
    for col in cols:
        mean_val = df[col].mean()
        std_val = df[col].std()
        z_score_norm[col] = (df[col] - mean_val) / std_val
    return z_score_norm

z_score_df = z_score_normalization(df, num_cols)

# 3. Decimal Scaling Normalization
def decimal_scaling_normalization(df, cols):
    dec_scale_norm = df.copy()
    for col in cols:
        j = len(str(abs(df[col].max())))
        dec_scale_norm[col] = df[col] / (10 ** j)
    return dec_scale_norm

dec_scale_df = decimal_scaling_normalization(df, num_cols)

# Display results
print("Original Data:")
print(df.head())
print("\nMin-Max Normalized Data:")
print(min_max_df.head())
print("\nZ-Score Normalized Data:")
print(z_score_df.head())
print("\nDecimal Scaling Normalized Data:")
print(dec_scale_df.head())
```
