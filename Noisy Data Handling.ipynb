```python
import pandas as pd
import numpy as np

# Sample dataset (Replace this with actual dataset loading)
data = [180.3086, 181.7717, 180.5674, 179.9005, 178.8166, 174.2771, 169.3205, 168.3252, 168.2057, 169.9276]
bins = 3  # Number of bins

# Equal Width Binning
min_val, max_val = min(data), max(data)
bin_width = (max_val - min_val) / bins
equal_width_bins = [[] for _ in range(bins)]
for value in data:
    bin_index = min(int((value - min_val) / bin_width), bins - 1)
    equal_width_bins[bin_index].append(value)

# Equal Frequency Binning
data_sorted = sorted(data)
bin_size = len(data) // bins
equal_frequency_bins = [data_sorted[i * bin_size:(i + 1) * bin_size] for i in range(bins)]

# Smoothing Functions
def smooth_mean(bins):
    return [[np.mean(bin_values)] * len(bin_values) for bin_values in bins]

def smooth_median(bins):
    return [[np.median(bin_values)] * len(bin_values) for bin_values in bins]

def smooth_bound(bins):
    return [[min(bin_values)] * (len(bin_values) // 2) + [max(bin_values)] * (len(bin_values) - len(bin_values) // 2) for bin_values in bins]

# Applying smoothing
smoothed_mean_eq_width = smooth_mean(equal_width_bins)
smoothed_mean_eq_freq = smooth_mean(equal_frequency_bins)

smoothed_median_eq_width = smooth_median(equal_width_bins)
smoothed_median_eq_freq = smooth_median(equal_frequency_bins)

smoothed_bound_eq_width = smooth_bound(equal_width_bins)
smoothed_bound_eq_freq = smooth_bound(equal_frequency_bins)

# Display results
def print_bins(title, bins):
    print(f"\n{title}")
    for i, bin_values in enumerate(bins):
        print(f"Bin {i+1}: {bin_values}")

print_bins("Equal Width Binning", equal_width_bins)
print_bins("Equal Frequency Binning", equal_frequency_bins)
print_bins("Smoothed Mean (Equal Width)", smoothed_mean_eq_width)
print_bins("Smoothed Mean (Equal Frequency)", smoothed_mean_eq_freq)
print_bins("Smoothed Median (Equal Width)", smoothed_median_eq_width)
print_bins("Smoothed Median (Equal Frequency)", smoothed_median_eq_freq)
print_bins("Smoothed Bound (Equal Width)", smoothed_bound_eq_width)
print_bins("Smoothed Bound (Equal Frequency)", smoothed_bound_eq_freq)
```
