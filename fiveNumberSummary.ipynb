```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('sample.csv')
df.drop(['Day','Month','Year'], axis=1, inplace=True)
df.drop(['Unnamed: 0'], axis=1, inplace=True)

column_name = "Close"
data = df[column_name].dropna().values  

# Calculate Five Number Summary
min_data = np.min(data)
Q1 = np.percentile(data, 25)  # First Quartile (25th percentile)
Q2 = np.percentile(data, 50)  # Median (50th percentile)
Q3 = np.percentile(data, 75)  # Third Quartile (75th percentile)
max_data = np.max(data)

# Print the Five Number Summary
print("\n📊 Five Number Summary for", column_name)
print(f"Minimum: {min_data}")
print(f"Q1 (25%): {Q1}")
print(f"Q2 (50%)(Median): {Q2}")
print(f"Q3 (75%): {Q3}")
print(f"Maximum: {max_data}")

# Create Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot(x=data, color="skyblue", width=0.3)
plt.title(f"Box Plot of {column_name}", fontsize=14)
plt.xlabel(column_name)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
